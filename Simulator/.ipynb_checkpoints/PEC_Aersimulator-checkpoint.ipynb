```python
Gamma_e = 2.8025 #(1/s.T)
Gamma_C13 = 1.0704*10**-3 #(1/s.T)

ti = 0 
tf = 10
itv = 1
t_total = np.arange(ti, tf+itv, itv)

BZ = [0, 50, 100, 200]  
R = ([np.array([0.07281108260423003, 0.3783374832738622, 0.36039647178489426]),\
      np.array([0.3640554130211498, -0.37833748327386224, -0.20594084101993956])])

#v,c,a用不同QC要改位置
#c、a用array後面的code才不用改
v = 2

c = np.zeros(len(R)).astype(int)
c[0] = 1
c[1] = 3

a = np.zeros(len(c)).astype(int)
a[0] = 0
a[1] = 4

'''
Position
0 - 1 - 2 - 3 - 4
'''
```

```python
cmax = max(c)
amax = max(a)
bits_len = max(v, max(amax, cmax)) + 1
bits_len
```

```python
R_z = np.zeros([len(BZ),len(t_total)]) 
theta = np.zeros([len(BZ),len(R),len(t_total)])
phi = np.zeros([len(BZ),len(R),len(t_total)])
Lambda = np.zeros([len(BZ),len(R),len(t_total)])
gamma = np.zeros([len(BZ),len(R),len(t_total)])

def normalize(h): #可亂打h
    a = h / np.sqrt(np.sum(h**2))
    return a

def A_z(h):
    n = 0.0198763
    d = (np.sqrt(np.sum(h**2)))**3
    a_x = (n/d)*(-3)*normalize(h)[2]*normalize(h)[0]
    a_y = (n/d)*(-3)*normalize(h)[2]*normalize(h)[1]
    a_z = (n/d)*(1-3*normalize(h)[2]**2)

    return np.array([a_x, a_y, a_z])

for bz in range(len(BZ)):
    omega_0 = (0,0,Gamma_C13*BZ[bz])
    for i in range(len(t_total)):         
        for r in range(len(R)):
            omega_1 = np.array([A_z(R[r])[0],A_z(R[r])[1],A_z(R[r])[2]+Gamma_C13*BZ[bz]])

            normalized_omega1 = normalize(omega_1)

            R_z[bz][i] = 2*np.pi*omega_0[2]*(itv*i+ti)

            theta[bz][r][i] = (2*(math.acos(np.sqrt(math.cos((2*np.pi*np.linalg.norm(omega_1)*(itv*i+ti))/2)**2 +
                                                  math.sin((2*np.pi*np.linalg.norm(omega_1)*(itv*i+ti))/2)**2*normalized_omega1[2]**2))))

            phi[bz][r][i] = ((-np.pi)/2
            +cmath.polar(complex(normalized_omega1[0],normalized_omega1[1]))[1]
            - cmath.polar(
                complex(
                    math.cos((2*np.pi*np.linalg.norm(omega_1)*(itv*i+ti))/2),
                    -math.sin((2*np.pi*np.linalg.norm(omega_1)*(itv*i+ti))/2)*normalized_omega1[2]))[1])

            Lambda[bz][r][i] = ((np.pi)/2
            -cmath.polar(complex(normalized_omega1[0],normalized_omega1[1]))[1]
            - 2*np.pi*Gamma_C13*BZ[bz]*(itv*i+ti)
            - cmath.polar(
                complex(
                    math.cos((2*np.pi*np.linalg.norm(omega_1)*(itv*i+ti))/2),
                    -math.sin((2*np.pi*np.linalg.norm(omega_1)*(itv*i+ti))/2)*normalized_omega1[2]))[1])

            gamma[bz][r][i] =((2*np.pi*Gamma_C13*BZ[bz]*(itv*i+ti))/2
            +cmath.polar(complex(
                    math.cos((2*np.pi*np.linalg.norm(omega_1)*(itv*i+ti))/2),
                    -math.sin((2*np.pi*np.linalg.norm(omega_1)*(itv*i+ti))/2)*normalized_omega1[2]))[1])
```

```python
#根據v,c,a的位置設定qubit的位元，ex:上面定a[0]=14,所以設15個qubit
qreg_q = QuantumRegister(bits_len, 'q')
creg_c = ClassicalRegister(bits_len, 'c')
qcl =[]

for bz in range(len(BZ)):
    omega_0 = (0,0,Gamma_C13*BZ[bz])    
    for i in range (len(t_total)):
        #IBM空白的電路圖
        circuit = QuantumCircuit(qreg_q, creg_c)

        #State preparation & polarization
        circuit.h(v)
        for j in range (len(c)):
            circuit.h(c[j])
            circuit.cx(c[j], a[j])
        #FID
        for j in range (len(c)):
            circuit.rz(R_z[bz][i], c[j])
            circuit.cu(theta[bz][j][i], phi[bz][j][i], Lambda[bz][j][i], gamma[bz][j][i], v, c[j])

        #QST
        circuit.h(v)
        #circuit.measure(v, v)

        qcl.append(circuit)
```
